```python
import numpy as np
import pandas as pd

def generate_knapsack_data(num_items, distribution='uniform', max_weight=100, max_value=100):
    if distribution == 'uniform':
        weights = np.random.randint(1, max_weight, num_items)
        values = np.random.randint(1, max_value, num_items)
    elif distribution == 'normal':
        weights = np.abs(np.random.normal(max_weight/2, max_weight/6, num_items)).astype(int)
        values = np.abs(np.random.normal(max_value/2, max_value/6, num_items)).astype(int)
    elif distribution == 'exponential':
        weights = np.random.exponential(max_weight/3, num_items).astype(int)
        values = np.random.exponential(max_value/3, num_items).astype(int)
    else:
        raise ValueError("Unsupported distribution type")
    
    # Ensure there are a few large items with high values
    num_high_value_items = max(1, num_items // 10)
    high_value_indices = np.random.choice(num_items, num_high_value_items, replace=False)
    weights[high_value_indices] = np.random.randint(max_weight // 2, max_weight, num_high_value_items)
    values[high_value_indices] = np.random.randint(max_value // 2, max_value, num_high_value_items)
    
    # Ensure there are many small items with high value/size ratios
    num_high_ratio_items = num_items - num_high_value_items
    high_ratio_indices = np.random.choice(np.setdiff1d(np.arange(num_items), high_value_indices), num_high_ratio_items, replace=False)
    weights[high_ratio_indices] = np.random.randint(1, max_weight // 4, num_high_ratio_items)
    values[high_ratio_indices] = np.random.randint(max_value // 2, max_value, num_high_ratio_items)
    
    return weights, values

num_instances = 50
num_items = 100
max_weight = 100
max_value = 100
distributions = ['uniform', 'normal', 'exponential']

# Prepare DataFrame to save to CSV
data_frames = []
for instance_id in range(num_instances):
    distribution = np.random.choice(distributions)
    weights, values = generate_knapsack_data(num_items, distribution, max_weight, max_value)
    capacity = np.random.randint(num_items * max_weight // 4, num_items * max_weight // 2)
    df = pd.DataFrame({
        'weights': weights,
        'values': values,
        'capacity': [capacity] * num_items,
        'distribution': [distribution] * num_items
    })
    df['instance_id'] = instance_id
    data_frames.append(df)

# Concatenate all DataFrames and save to CSV
result_df = pd.concat(data_frames, keys=range(num_instances))
result_df.to_csv('knapsack_datasets_suboptimal.csv')
```

```python
def read_knapsack_data_from_csv(csv_file):
    df = pd.read_csv(csv_file, index_col=[0,1])
    datasets = []
    for name, group in df.groupby(level=0):
        dataset = {
            'weights': group['weights'].tolist(),
            'values': group['values'].tolist(),
            'capacity': group['capacity'].iloc[0],
            'distribution': group['distribution'].iloc[0]
        }
        datasets.append(dataset)
    return datasets

# Usage
knapsack_datasets = read_knapsack_data_from_csv('knapsack_datasets.csv')
print(knapsack_datasets[0])  # Example: Print the first dataset
knapsack_datasets
```
